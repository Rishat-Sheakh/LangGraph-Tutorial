```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
```

```python
# Define State

# Define a Graph

# Define node to Graph

# add edges to Graph 

#Compile Graph

# Execute the graph 
```

```python
# Define State
class BMIState(TypedDict):
    weight_kg : float
    height_m : float
    bmi : float
    category : str

```

```python
def calculate_bmi(state: BMIState) -> BMIState:
    weight = state['weight_kg']
    height = state['height_m']

    bmi = weight/(height**2)

    state['bmi'] = round(bmi,2)

    return state


```

```python
def lebel_bmi(state: BMIState) -> BMIState:
    bmi = state['bmi']

    if bmi < 18.5:
        state["category"] = "Underweight"
    elif 18.5 <= bmi < 25:
        state["category"] = "Normal"
    elif 25 <= bmi < 30:
        state["category"] = "Overweight"
    else:
        state["category"] = "Obese"
        
    return state

    
```

```python
# Define a Graph

graph = StateGraph(BMIState)

# Define node to Graph
graph.add_node('calculate_bmi', calculate_bmi)
graph.add_node('lebel_bmi', lebel_bmi)

# add edges to Graph 
graph.add_edge(START, 'calculate_bmi')
graph.add_edge('calculate_bmi', 'lebel_bmi')
graph.add_edge('lebel_bmi', END)


```

```python
#Compile Graph
workflow = graph.compile()
```

```python
# exexute the graph

initial_state = {'weight_kg':80, 'height_m': 1.7}
final_state= workflow.invoke(initial_state)

print(final_state)
```
